```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1,2,3], [4,5,6])
plt.show()
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years, runs)
plt.show()
```

```python
plt.plot(years, runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly runs")
plt.show()
```

```python
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
 
plt.plot(years, kohli, label="Virat Kohli")
plt.plot(years, sehwag, label="Virendar Sehwag")

plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Performance Comparison")
plt.legend()
plt.show()
```

```python
plt.plot(years, kohli, 'ro--', label="Kohli") # red circles with dashed line
plt.plot(years, sehwag, 'g^:', label="Sehwag") # greeen triangles dotted
plt.legend()
```

```python
plt.plot(years, kohli, color='red', linestyle='--', label="Kohli")
plt.plot(years, sehwag, color='violet', linestyle=':', linewidth=4, label="Sehwag")
plt.plot(years, runs, color='blue', label="Tendulkar")
plt.legend()
```

```python
plt.plot(years, kohli, linewidth=3, label="Kohli")
plt.plot(years, sehwag, linewidth=2, label="Sehwag")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
